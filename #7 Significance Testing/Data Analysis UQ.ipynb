```python
# Import Required Libraries
import pandas as pd
import numpy as np
import os

# Set display options for better visualization
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
```

# Data Analysis - UQ Metrics

This notebook analyzes the uncertainty quantification metrics and predictions from all models.

```python
# Load ALL_UQ_METRICS.csv
uq_metrics_path = 'ALL_UQ_METRICS.csv'
df_uq_metrics = pd.read_csv(uq_metrics_path)

# Define the columns to display
columns_to_display = ['model_name', 'seed', 'mse', 'mae', 'rmse', 'mape', 'r2_score', 'picp', 'mpiw', 'winkler_score']

# Define model order
model_order = {'GRU': 0, 'LSTM': 1, 'TCN': 2, 'Transformer': 3}
seed_order = [42, 73, 1234, 31415, 271828183]

# Extract model name from model_name
df_uq_metrics['model'] = df_uq_metrics['model_name'].str.split('_').str[0].str.upper()

# Add numeric columns for proper sorting
df_uq_metrics['model_sort'] = df_uq_metrics['model'].map(model_order)
df_uq_metrics['seed_sort'] = df_uq_metrics['seed'].map({v: i for i, v in enumerate(seed_order)})

# Function to display scenario table with proper sorting and formatting
def display_scenario_table(df, scenario_name):
    print(f"\n{scenario_name} SCENARIO MODEL")
    df_filtered = df[df['model_name'].str.contains(scenario_name, case=False)][columns_to_display].copy()
    df_filtered['model'] = df_filtered['model_name'].str.split('_').str[0].str.upper()
    df_filtered['seed'] = df_filtered['model_name'].str.split('_').str[-1].astype(int)
    df_filtered['model_sort'] = df_filtered['model'].map(model_order)
    df_filtered['seed_sort'] = df_filtered['seed'].map({v: i for i, v in enumerate(seed_order)})
    df_filtered = df_filtered.sort_values(['model_sort', 'seed_sort']).drop(['model', 'seed', 'model_sort', 'seed_sort'], axis=1)
    df_filtered = df_filtered.reset_index(drop=True)
    
    # Format all numeric columns to 4 decimal places
    numeric_cols = df_filtered.select_dtypes(include=[np.number]).columns
    df_filtered_formatted = df_filtered.copy()
    for col in numeric_cols:
        df_filtered_formatted[col] = df_filtered_formatted[col].apply(lambda x: f'{x:.4f}')
    
    display(df_filtered_formatted)

# Display all three scenarios
display_scenario_table(df_uq_metrics, 'MCD')
display_scenario_table(df_uq_metrics, 'HLLLA')
display_scenario_table(df_uq_metrics, 'CQR')
```

```python
# Export data as plain text for copy-paste
def export_scenario_table_text(df, scenario_name):
    df_filtered = df[df['model_name'].str.contains(scenario_name, case=False)][columns_to_display].copy()
    df_filtered['model'] = df_filtered['model_name'].str.split('_').str[0].str.upper()
    df_filtered['seed'] = df_filtered['model_name'].str.split('_').str[-1].astype(int)
    df_filtered['model_sort'] = df_filtered['model'].map(model_order)
    df_filtered['seed_sort'] = df_filtered['seed'].map({v: i for i, v in enumerate(seed_order)})
    df_filtered = df_filtered.sort_values(['model_sort', 'seed_sort']).drop(['model', 'seed', 'model_sort', 'seed_sort'], axis=1)
    df_filtered = df_filtered.reset_index(drop=True)
    
    # Format numeric columns to 4 decimal places
    numeric_cols = df_filtered.select_dtypes(include=[np.number]).columns
    for col in numeric_cols:
        df_filtered[col] = df_filtered[col].apply(lambda x: f'{x:.4f}')
    
    # Print as plain text
    print(f"\n{'='*120}")
    print(f"{scenario_name} SCENARIO MODEL")
    print(f"{'='*120}")
    print(df_filtered.to_string(index=False))
    print(f"{'='*120}\n")

# Export all three scenarios
export_scenario_table_text(df_uq_metrics, 'MCD')
export_scenario_table_text(df_uq_metrics, 'HLLLA')
export_scenario_table_text(df_uq_metrics, 'CQR')
```

```python
# Function to compute mean ± std grouped by model
def compute_mean_std_by_model(df, scenario_name):
    print(f"\n{scenario_name} SCENARIO MODEL - Mean ± Std by Model")
    df_filtered = df[df['model_name'].str.contains(scenario_name, case=False)].copy()
    df_filtered['model'] = df_filtered['model_name'].str.split('_').str[0].str.upper()
    
    # Select only numeric columns (excluding model columns)
    numeric_cols = [col for col in columns_to_display if col not in ['model_name', 'seed']]
    
    # Group by model and compute mean and std
    grouped_stats = []
    for model in ['GRU', 'LSTM', 'TCN', 'TRANSFORMER']:
        model_data = df_filtered[df_filtered['model'] == model][numeric_cols]
        if len(model_data) > 0:
            stats_row = {'Model': model}
            for col in numeric_cols:
                mean_val = model_data[col].mean()
                std_val = model_data[col].std()
                stats_row[col] = f'{mean_val:.4f} ± {std_val:.4f}'
            grouped_stats.append(stats_row)
    
    df_stats = pd.DataFrame(grouped_stats)
    display(df_stats)

# Display mean ± std for all three scenarios
compute_mean_std_by_model(df_uq_metrics, 'MCD')
compute_mean_std_by_model(df_uq_metrics, 'HLLLA')
compute_mean_std_by_model(df_uq_metrics, 'CQR')
```

```python
# Bottom 10% based on Winkler Score (lower is better)
print("Bottom 10% of Results Based on Winkler Score (Lower is Better)")
print("="*100)

# Calculate 10th percentile of winkler_score
percentile_10 = df_uq_metrics['winkler_score'].quantile(0.10)
print(f"10th Percentile Winkler Score: {percentile_10:.4f}\n")

# Filter bottom 10%
bottom_10_percent = df_uq_metrics[df_uq_metrics['winkler_score'] <= percentile_10][columns_to_display].copy()

# Sort by winkler_score ascending (best first)
bottom_10_percent = bottom_10_percent.sort_values('winkler_score', ascending=True).reset_index(drop=True)

# Format numeric columns to 4 decimal places
numeric_cols = bottom_10_percent.select_dtypes(include=[np.number]).columns
df_bottom10_formatted = bottom_10_percent.copy()
for col in numeric_cols:
    df_bottom10_formatted[col] = df_bottom10_formatted[col].apply(lambda x: f'{x:.4f}')

display(df_bottom10_formatted)
```
